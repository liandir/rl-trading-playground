```
import numpy as np
import matplotlib.pyplot as plt
```

```
alpha = 0.5
beta = 2

x = np.linspace(0,1,1001)
f = x**(alpha-1) * (1-x)**(beta-1) # 

fig, ax = plt.subplots(figsize=(12,4))
ax.plot(x,f)
ax.set_title(f"Beta distribution with alpha={alpha} and beta={beta}")
ax.set_xlabel("x")
ax.set_ylabel("f(x)")
ax.grid(alpha=0.3)
plt.show()
```
