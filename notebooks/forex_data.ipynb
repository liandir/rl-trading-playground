```
import os, sys
sys.path.append("..")
from datetime import datetime, timezone

import matplotlib.pyplot as plt

from forex import FOREX
```

```
UTC = timezone.utc

fx = FOREX(
    data_path="../data",
    assets=["USDEUR", "JPYEUR", "GBPEUR", "CHFEUR", "AUDEUR"],
)

data = fx.get(
    start=datetime(2026, 1, 1, tzinfo=UTC),
    end=datetime(2026, 4, 12, tzinfo=UTC),
    interval="1min",
)

print(data)
```

```
def read_forex_data(file_path):
    data = []

    with open(file_path, 'r') as f:
        lines = f.readlines()

    headers = lines[0].strip().split(',')
    print(headers)

    for line in lines[1:]:
        parts = line.strip().split(',')
        t = datetime.strptime(parts[0], "%d.%m.%Y %H:%M:%S.%f UTC")
        open_ = 1/float(parts[1])
        high = 1/float(parts[2])
        low = 1/float(parts[3])
        close = 1/float(parts[4])
        volume = float(parts[5]) if parts[5] else 0.0
        data.append({
            "time": t,
            "open": open_,
            "high": high,
            "low": low,
            "close": close,
            "volume": volume
        })

    return headers, data
```

```
def get_field(data, field):
    return [entry[field] for entry in data]
```
