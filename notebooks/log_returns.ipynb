```
# Demonstrate why log-returns correspond to continuous compounding
# and how log(1+R) approximates R for small returns.

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from math import exp, log
# from caas_jupyter_tools import display_dataframe_to_user

np.random.seed(0)
```

```
# 1) Compare log(1+R) to R over a small range
R = np.linspace(-0.3, 0.3, 601)  # +/-30% moves
log1p_R = np.log1p(R)
taylor1 = R
taylor2 = R - 0.5 * R**2
taylor3 = R - 0.5 * R**2 + (R**3)/3

plt.figure(figsize=(10, 4), dpi=100)
plt.plot(R, log1p_R, label="log(1+R)")
plt.plot(R, taylor1, label="1st-order ~ R")
plt.plot(R, taylor2, label="2nd-order ~ R - R^2/2")
plt.plot(R, taylor3, label="3rd-order ~ R - R^2/2 + R^3/3")
plt.axhline(0, linewidth=0.8, color="black")
plt.axvline(0, linewidth=0.8, color="black")
plt.title("log(1+R) vs. Taylor Approximations")
plt.xlabel("Simple return R")
plt.ylabel("Value")
plt.grid(alpha=0.3)
plt.legend()
plt.show()
```

```
# 2) Simulate a return path and compare compounding
T = 250  # trading days
mu_daily = 0.0005   # 0.05% per day drift
sigma_daily = 0.01  # 1% daily vol
r_simple = np.random.normal(mu_daily, sigma_daily, size=T)  # simple returns ~ N(mu, sigma)

# Wealth paths starting from V0=1
V0 = 1.0
V_discrete = V0 * np.cumprod(1 + r_simple)  # discrete compounding with simple returns
cum_log = np.cumsum(np.log1p(r_simple))     # cumulative log-returns
V_cont = V0 * np.exp(cum_log)               # continuous compounding using log-returns (exact by definition)

plt.figure(figsize=(7, 4))
plt.plot(V_discrete, label=r"Discrete compounding $\left(\prod(1+R_t)\right)$")
plt.plot(V_cont, label=r"Continuous compounding $\left(exp \left(\sum log(1+R_t)\right)\right)$", linestyle="--")
plt.title("Wealth Paths: Discrete vs. Continuous Compounding")
plt.xlabel("Time step")
plt.ylabel("Portfolio value (V)")
plt.grid(alpha=0.3)
plt.legend()
plt.show()
```

```
# 3) Show tracking error statistics
tracking_error = V_cont - V_discrete
summary = pd.DataFrame({
    "final_V_discrete": [V_discrete[-1]],
    "final_V_cont": [V_cont[-1]],
    "final_diff": [tracking_error[-1]],
    "max_abs_diff": [np.max(np.abs(tracking_error))]
})
summary
```
